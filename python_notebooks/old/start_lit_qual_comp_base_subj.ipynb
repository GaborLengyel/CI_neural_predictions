```python
def convert_to_velocity(speed, angle_degrees):
    # Convert the angle from degrees to radians
    angle_radians = np.radians(angle_degrees)
    
    # Calculate velocity components
    v_x = speed * np.cos(angle_radians)
    v_y = speed * np.sin(angle_radians)
    
    return v_x, v_y

def velocity_to_speed_angle(v_x, v_y):
    # Compute the speed (magnitude of the vector)
    speed = np.sqrt(v_x**2 + v_y**2)
    
    # Compute the angle in radians and convert to degrees
    angle_radians = np.arctan2(v_y, v_x)
    angle_degrees = np.degrees(angle_radians)
    
    return speed, angle_degrees

def subtract_angles(angle1, angle2):
    result = (angle1 - angle2) % 360
    if result > 180:
        result -= 360
    return result
```

```python
sd= np.linspace(-180,180,361)
Sspd = np.linspace(0,5,5)
Cspd = np.linspace(0,5,5)
diffS = np.zeros((len(sd),len(Sspd),len(Cspd)))
diffA = np.zeros((len(sd),len(Sspd),len(Cspd))) 
for i,sspd in enumerate(Sspd):
    for j,cspd in enumerate(Cspd):
        vc_x, vc_y = convert_to_velocity(sspd, 0)
        for k,s in enumerate(sd):
            vs_x, vs_y = convert_to_velocity(cspd, s)
            # Calculate the difference of the vectors
            v_diff_x = vc_x - vs_x
            v_diff_y = vc_y - vs_y
            # Convert the difference back to speed and angle
            speed_diff, angle_diff = velocity_to_speed_angle(v_diff_x, v_diff_y)

            diffS[k,i,j] = speed_diff
            diffA[k,i,j] = angle_diff
```
